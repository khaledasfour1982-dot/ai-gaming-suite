```python
import pandas as pd
import numpy as np
import json

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 200)

CLEAN_BASE_PATH = "../logs_clean_base.parquet"
df_base = pd.read_parquet(CLEAN_BASE_PATH)

print("Clean base data shape:", df_base.shape)
display(df_base.head())
```

```python
print("Columns in df_base:")
print(df_base.columns.tolist())

print("\nMissing values per column:")
display(df_base.isna().sum())

if "outcome" in df_base.columns:
    print("\nOutcome distribution (clean base):")
    print(df_base["outcome"].value_counts(dropna=False))

if "move_index" in df_base.columns:
    print("\nmove_index summary:")
    print(df_base["move_index"].describe())
```

```python
df = df_base.copy()

# Sort within each game by time / move_index (should already be sorted, but we enforce it)
sort_cols = ["game_session_id"]
if "logged_at" in df.columns:
    sort_cols.append("logged_at")
elif "move_index" in df.columns:
    sort_cols.append("move_index")

df = df.sort_values(sort_cols).reset_index(drop=True)

# final_outcome = last outcome per game
df["final_outcome"] = df.groupby("game_session_id")["outcome"].transform("last")

print("final_outcome distribution:")
print(df["final_outcome"].value_counts(dropna=False))
```

```python
# Winner per game = player who makes the last move in games with WIN/LOSS
winner_by_game = (
    df[df["final_outcome"].isin(["WIN", "LOSS"])]
    .groupby("game_session_id")["played_by"]
    .last()
)

df["game_winner"] = df["game_session_id"].map(winner_by_game)


def win_pov(row):
    if row["final_outcome"] == "DRAW":
        return 0.5
    if row["final_outcome"] not in ("WIN", "LOSS"):
        return np.nan
    if pd.isna(row["played_by"]) or pd.isna(row["game_winner"]):
        return np.nan
    return 1.0 if row["played_by"] == row["game_winner"] else 0.0


df["win_label_pov"] = df.apply(win_pov, axis=1)

print("win_label_pov distribution (including NaN):")
print(df["win_label_pov"].value_counts(dropna=False))
```

```python
before = df.shape[0]
df = df[df["win_label_pov"].notna()].copy()
after = df.shape[0]

print(f"Removed {before - after} rows with NaN win_label_pov.")
print("Shape after target filtering:", df.shape)

print("\nwin_label_pov distribution (final):")
print(df["win_label_pov"].value_counts())
```

```python
played_map = {"X": 0, "O": 1}
df["played_by_encoded"] = df["played_by"].map(played_map)

print("played_by vs played_by_encoded:")
display(df[["played_by", "played_by_encoded"]].head())

print("Missing in played_by_encoded:", df["played_by_encoded"].isna().sum())
```

```python
if "ai_difficulty" in df.columns:
    df["ai_difficulty"] = df["ai_difficulty"].astype("category")
    print("ai_difficulty categories:", df["ai_difficulty"].cat.categories.tolist())
else:
    print("No `ai_difficulty` column found – skipping.")
```

```python
def parse_board_to_vector(board_value):
    """
    board_value ma postać np.:
      [["", "", ""],
       ["", "X", ""],
       ["", "", ""]]

    albo string z JSON-em takiej struktury.

    Encoding:
    - "" / None      -> 0
    - "X"            -> 1
    - "O"            -> 2
    """
    if board_value is None or (isinstance(board_value, float) and pd.isna(board_value)):
        return None

    if isinstance(board_value, str):
        s = board_value.strip()
        try:
            board = json.loads(s)
        except Exception:
            return None
    else:
        board = board_value

    if not isinstance(board, list) or len(board) == 0:
        return None
    if not isinstance(board[0], list):
        return None

    n_rows = len(board)
    n_cols = len(board[0])
    board_size = n_rows * n_cols

    value_map = {
        None: 0,
        "": 0,
        " ": 0,

        "X": 1,
        "CROSS": 1,
        "PLAYER_X": 1,

        "O": 2,
        "CIRCLE": 2,
        "PLAYER_O": 2,
    }

    vec = []
    for r in range(n_rows):
        row_list = board[r]
        if not isinstance(row_list, list):
            return None
        if len(row_list) < n_cols:
            row_list = row_list + [None] * (n_cols - len(row_list))
        for c in range(n_cols):
            val = row_list[c]
            mapped = value_map.get(val, 0)
            vec.append(mapped)

    if len(vec) != board_size:
        return None

    return vec
```

```python
board_cols = []

if "board" in df.columns:
    board_vectors = df["board"].apply(parse_board_to_vector)

    valid_lengths = board_vectors.dropna().map(len)

    if len(valid_lengths) == 0:
        print("No valid `board` vectors parsed – check logging format.")
    else:
        board_size = valid_lengths.mode().iloc[0]
        print(f"Detected board size (number of cells): {board_size}")

        for i in range(board_size):
            col_name = f"board_{i}"
            board_cols.append(col_name)
            df[col_name] = board_vectors.apply(
                lambda v, idx=i: v[idx] if (isinstance(v, list) and len(v) == board_size) else np.nan
            )

        print("Created board columns:", board_cols)
        display(df[board_cols].head())
else:
    print("No `board` column found – skipping board feature engineering.")
```

```python
if board_cols:
    complete_mask = df[board_cols].notna().all(axis=1)
    print("Rows with complete board features:", int(complete_mask.sum()))
    print("Total rows:", df.shape[0])
    print("Share with complete boards: {:.2%}".format(complete_mask.mean()))
else:
    print("No board_* columns were created.")
```

```python
board_values = df[board_cols].fillna(0)

df["board_X_count"] = (board_values == 1).sum(axis=1)
df["board_O_count"] = (board_values == 2).sum(axis=1)
df["board_empty_count"] = (board_values == 0).sum(axis=1)

print(
    "Board feature summaries:\n"
    "X count:\n", df["board_X_count"].describe(), "\n\n",
    "O count:\n", df["board_O_count"].describe(), "\n\n",
    "Empty count:\n", df["board_empty_count"].describe()
)

display(df[board_cols + ["board_X_count", "board_O_count", "board_empty_count"]].head())
```

```python
target_col = "win_label_pov"

id_cols = ["game_session_id"]
id_cols = [c for c in id_cols if c in df.columns]

feature_cols = [
    "move_index",
    "played_by_encoded",
    "ai_difficulty",
    "board_X_count",
    "board_O_count",
    "board_empty_count",
] + board_cols

feature_cols_existing = [c for c in feature_cols if c in df.columns]

cols_to_keep = id_cols + feature_cols_existing + [target_col]
cols_to_keep = list(dict.fromkeys(cols_to_keep))

print("\nColumns that will be kept for modeling (all others are dropped):")
print(cols_to_keep)

df_model = df[cols_to_keep].copy()

print("Modeling dataset shape:", df_model.shape)
display(df_model.head())

MODEL_READY_PATH = "../logs_model_ready.parquet"
df_model.to_parquet(MODEL_READY_PATH)

print(f"Saved modeling dataset to: {MODEL_READY_PATH}")
```
